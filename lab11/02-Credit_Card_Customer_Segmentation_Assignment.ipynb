## Feature Scaling

K-Means is a distance-based algorithm. Therefore, feature scaling is very important.

The features in this dataset have very different ranges. For example, `BALANCE`, `PURCHASES`, and `CREDIT_LIMIT` may have large values, while frequency columns are between 0 and 1.

**Use StandardScaler to scale the data. Save the scaled data in a variable called `X_scaled`.**

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

## Choosing K Intuitively

Choosing K is one of the most difficult parts of K-Means.

Since this dataset has many features, it is not easy to visually see the clusters directly.

However, we can still compare different K values using the elbow method and silhouette score.

## Elbow Method

**Create a loop that fits K-Means models for K values from 1 to 10. Save the inertia values in a list called `inertia_values`.**

```python
inertia_values = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)
```

**Plot the elbow curve.**

```python
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertia_values, marker='o', color='steelblue')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method - Choosing Optimal K')
plt.xticks(range(1, 11))
plt.tight_layout()
plt.show()
```

**Output Interpretation**

Look at the elbow curve and try to identify where the decrease in inertia starts to slow down.

That point can suggest a reasonable value for K.

## Silhouette Score

The silhouette score helps evaluate how well-separated the clusters are.

**Create a loop that calculates the silhouette score for K values from 2 to 10. Save the scores in a list called `silhouette_scores`.**

```python
silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)
```

**Plot the silhouette scores.**

```python
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_scores, marker='o', color='coral')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Different K Values')
plt.xticks(range(2, 11))
plt.tight_layout()
plt.show()
```

**Create a table showing each K value and its silhouette score.**

```python
silhouette_df = pd.DataFrame({
    'K': range(2, 11),
    'Silhouette Score': silhouette_scores
})
silhouette_df
```

**Output Interpretation**

A higher silhouette score usually means better clustering.

However, do not rely only on the highest value. Also consider whether the chosen K makes sense for customer segmentation.

## Create the Final K-Means Model

**Based on the elbow curve and silhouette scores, choose a final K value. Then train a final K-Means model.**

Use `random_state=42` and `n_init=10`.

```python
# K=3 is chosen based on the elbow curve showing a bend around K=3
# and K=3 giving a reasonable silhouette score with meaningful business segments
final_kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
final_kmeans.fit(X_scaled)
```

**Add the final cluster labels to the original dataframe in a new column called `Cluster`.**

```python
df['Cluster'] = final_kmeans.labels_
```

**Check the first five rows after adding the cluster labels.**

```python
df.head()
```

## Cluster Analysis

Now we need to understand what each cluster means.

## Visualizing the Final Clusters

Since the dataset has many features, we will use PCA to reduce the data into two components only for visualization.

This visualization does not replace the original clustering. It only helps us see the clusters in a 2D plot.

**Use PCA with 2 components and plot the clusters.**

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(9, 6))
colors = ['steelblue', 'coral', 'green']
for cluster in sorted(df['Cluster'].unique()):
    mask = df['Cluster'] == cluster
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1],
                label=f'Cluster {cluster}',
                alpha=0.4, s=20, color=colors[cluster])

plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.title('Customer Clusters Visualized with PCA')
plt.legend()
plt.tight_layout()
plt.show()
```

**Output Interpretation**

The PCA plot gives a simplified 2D view of the clusters.

If the clusters are not perfectly separated, that is normal because the original dataset has many features and the plot only shows two compressed dimensions.

## Final Questions

**1. Why is this an unsupervised learning problem?**

This is an unsupervised learning problem because the dataset does not contain a predefined target label or output variable. There is no column that tells us which group a customer belongs to. The algorithm must discover the hidden patterns and group customers on its own, based only on the input features.

---

**2. Why did we remove the `CUST_ID` column?**

The `CUST_ID` column is just a unique identifier for each customer. It carries no information about customer behavior. If we kept it, the clustering algorithm might use it as a feature, which would produce meaningless results. We only want the model to group customers based on their actual behavioral data.

---

**3. Which columns had missing values?**

Two columns had missing values:
- `CREDIT_LIMIT`: 1 missing value
- `MINIMUM_PAYMENTS`: 313 missing values

---

**4. How did you handle the missing values?**

We used mean imputation. Each missing value was replaced with the mean of its column. This approach preserves all rows in the dataset and avoids reducing the sample size, which is important when we want to cluster all available customers.

---

**5. Why is scaling important before applying K-Means?**

K-Means calculates distances between data points to form clusters. If features have very different scales, features with large numerical ranges (like `BALANCE` or `PURCHASES` which can reach thousands) will dominate the distance calculation over features with small ranges (like frequency columns between 0 and 1). Scaling ensures every feature contributes equally to the clustering.

We chose **K = 3**.

- **Elbow Method**: The inertia drops steeply from K=1 to K=3, and after K=3, the decrease becomes much smaller. This bend in the curve at K=3 suggests it is a reasonable point where adding more clusters brings diminishing returns.
- **Silhouette Score**: K=3 has a relatively high silhouette score compared to higher K values, and it produces clusters that are interpretable from a business perspective.
- **Business Logic**: Three segments align well with common credit card marketing strategies: low-activity customers, active purchasers, and cash-advance-dependent customers.
